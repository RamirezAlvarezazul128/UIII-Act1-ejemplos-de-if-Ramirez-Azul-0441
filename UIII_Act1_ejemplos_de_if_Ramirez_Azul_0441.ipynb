#### Ejemplo de if

```python
# 1. Ejemplo de if simple (con operador relacional)
# La estructura if evalúa una condición y, si es verdadera, ejecuta el bloque de código indentado
edad = 18 # Asigna el valor 18 a la variable edad
if edad >= 18: # Compara si edad es mayor o igual a 18 [8]
    print("Eres mayor de edad.") # Se ejecuta solo si la comparación es True [1]
```

#### Ejemplo de if else

```python
#2. Ejemplo de if-else (con operadores aritméticos y relacionales)
#La estructura if-else permite manejar dos alternativas: una si la condición se cumple y otra si es falsa

numero = 10 # Define la variable numero
if numero % 2 == 0: # Usa el operador módulo (%) para obtener el resto y == para comparar [3, 5]
    print("El número es par.") # Se ejecuta si el resto de dividir por 2 es cero [11]
else: # Bloque opcional para el caso falso [1]
    print("El número es impar.") # Se ejecuta si el resto no es cero [11]
```

#### Ejemplo de if-elif

```python
#3. Ejemplo de if-elif-else (con operadores lógicos y relacionales)
#Esta estructura se utiliza cuando hay más de dos caminos posibles
puntaje = int(input("ingresa una calificacion"))
puntaje = 85 # Define una variable de calificación
if puntaje >= 90: # Comprueba si el puntaje es sobresaliente [13]
    print("Calificación: A") # Resultado para puntajes muy altos
elif puntaje >= 80 and puntaje < 90: # Combina comparaciones usando el operador lógico 'and' [7]
    print("Calificación: B") # Se ejecuta si ambas condiciones son True [7]
else: # Caso por defecto si ninguna anterior se cumple [1]
    print("Calificación: C o inferior") # Resultado para el resto de los casos
```

#### Ejemplo combinado(operadores logicos y raicionales en un rango)

```python
#4. Ejemplo complejo (Aritméticos + Lógicos + Relacionales)
#Este ejemplo verifica si un año es bisiesto, lo cual requiere una lógica de múltiples pasos
anio = 2024 # Define el año a verificar
# Un año es bisiesto si es divisible por 4 y (no por 100 o sí por 400) [8, 14]
if (anio % 4 == 0 and anio % 100 != 0) or (anio % 400 == 0):
    print(f"{anio} es un año bisiesto.") # Mensaje si cumple la regla lógica [8]
else:
    print(f"{anio} no es un año bisiesto.") # Mensaje si no la cumple [8]
```
